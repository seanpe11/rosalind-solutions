# [Consensus and Profile](https://rosalind.info/problems/cons/)
My general idea to solve this is to think in terms of parallel strings. Since the profile matrix takes the count of each, we can thus use the profile matrix to generate the consensus string by just taking the greater of the hashmap for the given column. Best datastructure to process the strings and to get each count of ACTG was to make a HashMap with a key of <columnNo + "A" | "C" | "T" | "G"> and value of an int which represents the count of the character in that column.

```python
def cons_solution(filename="./inputs/cons_sample.txt"):
    inp = open(filename)
    strings = []
    to_add = ""
    for x in inp:
        if (x.startswith('>')):
            strings.append(to_add)
            to_add = ""
            continue
        else:
            to_add += x.strip()
    strings.pop(0)

    print(strings)
    # Strings are equal length, so just go column by column and get counts
    profileM = {} # dict makes the most sense here, we'll use column + char as index
    chars = ['A', 'C', 'G', 'T']
    for x in range(len(strings[0])): # init key value pairs
        for c in chars:
            profileM[str(x)+c] = 0
    # now traverse DNA strings and update counts per column
    for x in range(len(strings[0])):
        for y in range(len(strings)):
            profileM[str(x) + (strings[y][x])] +=1

    final = ""
    for x in range(len(strings[0])):
        vals = [profileM[str(x) + c] for c in chars]
        idx = vals.index(max(vals))
        final += chars[idx]
    return final
        
print(cons_solution())
print(cons_solution("./inputs/cons.txt"))
```
